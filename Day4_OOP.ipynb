# Day 4: Object-Oriented Programming (OOP) - Theory Notes

**1. Classes and Objects**
A **Class** is a blueprint or template for creating objects. It defines the properties (attributes) and behaviors (methods) that objects created from the class will have. An **Object** is an instance of a class. It is a concrete entity that holds actual data.
* *Difference:* A class is a logical construct (like an architectural blueprint), while an object is a physical reality (like the actual house built from the blueprint).

**2. Constructors (`__init__`)**
A constructor is a special method automatically called when a new object of a class is created. In Python, it is named `__init__`. Its primary purpose is to initialize the object's attributes.

**3. Instance vs. Class Variables**
* **Instance Variables:** Tied to a specific object. Each object has its own copy (e.g., `self.name`).
* **Class Variables:** Shared across all instances of a class. Defined directly inside the class but outside any methods.

**4. The Four Pillars of OOP**
* **Encapsulation:** Wrapping data (variables) and methods into a single unit (class) and restricting direct access to some of the object's components (using private variables).
* **Inheritance:** A mechanism where a new class (child) inherits properties and methods from an existing class (parent), promoting code reusability.
* **Polymorphism:** The ability of different objects to respond to the same method call in their own specific way (e.g., both `Car` and `Bike` have a `start()` method, but they behave differently).
* **Abstraction:** Hiding complex implementation details and showing only the essential features of an object (often achieved using abstract classes and methods).

**5. Overriding vs. Overloading**
* **Method Overriding:** When a child class provides a specific implementation for a method that is already defined in its parent class.
* **Method Overloading:** Having multiple methods with the same name but different parameters within the same class (Note: Python does not support traditional method overloading directly; it is usually handled via default arguments).

**6. Access Modifiers & `super()`**
* **Public:** Accessible from anywhere.
* **Protected (`_var`):** Accessible within the class and its subclasses.
* **Private (`__var`):** Accessible only within the class itself.
* **`super()`:** A built-in function used to call methods (most commonly the constructor) from a parent class.

```python
# Practice 1: Student Management System
class Student:
    # Constructor
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.marks = marks
    
    def calculate_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 75:
            return "B"
        elif self.marks >= 60:
            return "C"
        else:
            return "F"
            
    def display(self):
        print(f"Student: {self.name} | Roll No: {self.roll} | Marks: {self.marks} | Grade: {self.calculate_grade()}")

# Execution
student1 = Student("Arya", 101, 85)
student2 = Student("Kavya", 102, 92)

student1.display()
student2.display()
```

```python
# Practice 2: Bank Account System (Encapsulation)
class BankAccount:
    def __init__(self, initial_balance):
        # The double underscore makes this a private variable
        self.__balance = initial_balance 
        
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: ₹{amount}")
            
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew: ₹{amount}")
        else:
            print("Insufficient funds or invalid amount.")
            
    def get_balance(self):
        return f"Current Balance: ₹{self.__balance}"

# Execution
account = BankAccount(1000)
account.deposit(500)
account.withdraw(200)
print(account.get_balance())

# Trying to access the private variable directly will cause an error:
# print(account.__balance)  # This would fail, demonstrating encapsulation
```

```python
# Practice 3: Employee Management (Inheritance)
class Employee:
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id
        
    def work(self):
        print(f"{self.name} is working on general tasks.")

# Manager inherits from Employee
class Manager(Employee):
    def __init__(self, name, emp_id, department):
        # super() is used to call the parent class's constructor
        super().__init__(name, emp_id)
        self.department = department
        
    def manage(self):
        print(f"{self.name} is managing the {self.department} team.")

# Execution
emp = Employee("Rahul", "E001")
mgr = Manager("Sneha", "M001", "Data Science")

emp.work()
mgr.work()   # Inherited method
mgr.manage() # Child-specific method
```

```python
# Practice 4: Vehicle System (Polymorphism & Method Overriding)
class Vehicle:
    def start(self):
        print("Generic vehicle starts.")
        
class Car(Vehicle):
    # Overriding the parent's start method
    def start(self):
        print("Car starts with a key turn.")
        
class Bike(Vehicle):
    # Overriding the parent's start method
    def start(self):
        print("Bike starts with a self-start button.")

# Execution
vehicles = [Vehicle(), Car(), Bike()]

# Demonstrating Polymorphism: 
# The same method name 'start()' behaves differently depending on the object
for v in vehicles:
    v.start()
```

```python
# Practice 5: Shape Area Calculator (Abstraction)
from abc import ABC, abstractmethod

# Abstract Base Class
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.l = length
        self.b = breadth
        
    # We must implement the abstract method, otherwise Rectangle cannot be instantiated
    def area(self):
        return self.l * self.b

# Execution
rect = Rectangle(10, 5)
print(f"Area of Rectangle: {rect.area()}")

# Trying to instantiate an abstract class directly will cause an error:
# shape = Shape() This would fail, demonstrating abstraction
```

```python
# Practice 6: Library Management System (Mini Project)

class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_issued = False

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

class Library:
    def __init__(self):
        self.books = []
        
    def add_book(self, book):
        self.books.append(book)
        print(f"Book added: '{book.title}'")
        
    def display_available_books(self):
        print("\n    Available Books    ")
        available = [b.title for b in self.books if not b.is_issued]
        if available:
            for title in available:
                print(f"- {title}")
        else:
            print("No books currently available.")
            
    def issue_book(self, title, member):
        for book in self.books:
            if book.title == title and not book.is_issued:
                book.is_issued = True
                member.borrowed_books.append(book)
                print(f"\n'{title}' has been issued to {member.name}.")
                return
        print(f"\nSorry, '{title}' is either unavailable or doesn't exist.")
        
    def return_book(self, title, member):
        for book in member.borrowed_books:
            if book.title == title:
                book.is_issued = False
                member.borrowed_books.remove(book)
                print(f"\n{member.name} has returned '{title}'.")
                return
        print(f"\n{member.name} does not have '{title}'.")

#     Execution / Testing the Mini Project
my_library = Library()

# Add Books
my_library.add_book(Book("Python Crash Course", "Eric Matthes"))
my_library.add_book(Book("Data Science from Scratch", "Joel Grus"))

# Display
my_library.display_available_books()

# Create Member and Issue
student = Member("Arya")
my_library.issue_book("Python Crash Course", student)

# Display again to see it was removed
my_library.display_available_books()

# Return Book
my_library.return_book("Python Crash Course", student)
my_library.display_available_books()
```
